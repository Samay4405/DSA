```python
#  HASHABLE / IMMUTABLE
#       strings
#       integers
#       tuples
#  UNHASHABLE / MUTABLE
#       lists
#       dictionaries (they are hashmap themselves)
#       sets    
```

```python
# Hashsets

s = set()
print(s)
```

```python
# Add item into Set - O(1)
s.add(1)
s.add(2)
s.add(3)

print(s)
```

```python
# Lookup if item in set - O(1)
if 1 not in s:
  print(True)
```

```python
# Remove item from set - O(1)
s.remove(3)

print(s)
```

```python
# Set construction - O(S) - S is the length of the string
string = 'aaaaaaabbbbbbbbbbbccccccccceeeeeeeee'
sett = set(string)

sett
```

```python
# Loop over items in set - O(n)
for x in s:
  print(x)
```

```python
# Hashmaps - Dictionaries

d = {'greg': 1, 'steve': 2, 'rob': 3}

print(d)
```

```python
# Add key:val in dictionary: O(1)
d['arsh'] = 4

print(d)
```

```python
# Check for presence of key in dictionary: O(1)
if 'greg' in d:
  print(True)
```

```python
# Check the value corresponding to a key in the dictionary: O(1)
print(d['greg'])
```

```python
# Loop over the key:val pairs of the dictionary: O(n)
for key, val in d.items():
  print(f'key {key}: val {val}')
```

```python
# Defaultdict - A dictionary that returns a default value if the key is not present in the dictionary. It is a subclass of the built-in dict class.

from collections import defaultdict

default = defaultdict(list)

default[2]
```

```python
# Counter - A dictionary subclass for counting hashable objects. It is a collection where elements are stored as dictionary keys and their counts are stored as dictionary values.
from collections import Counter

counter = Counter(string)

print(counter)
```
